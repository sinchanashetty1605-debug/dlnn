```python
import pandas as pd
import numpy as np
dataset=pd.read_csv("/content/housing.csv")
dataset.head()
```

```python
dataset.shape
```

```python
from sklearn import preprocessing
import numpy as np

# Select only numerical columns for scaling, EXCLUDING the target variable
x = dataset.drop('median_house_value', axis=1).select_dtypes(include=np.number).copy()

# Handle missing values in numerical columns (e.g., total_bedrooms has some NaNs)
x.fillna(x.mean(), inplace=True)

min_max_scalar=preprocessing.MinMaxScaler()
x_scale = min_max_scalar.fit_transform(x)
print(x_scale)
```

```python
from sklearn.model_selection import train_test_split
from sklearn import preprocessing

y_unscaled = dataset['median_house_value'] # Keep original y for inverse transformation if needed

# Scale the target variable 'y'
min_max_scalar_y = preprocessing.MinMaxScaler()
y_scale = min_max_scalar_y.fit_transform(y_unscaled.values.reshape(-1, 1))

x_train,x_test,y_train,y_test=train_test_split(x_scale,y_scale,test_size=0.30,random_state=0)
```

```python
from tensorflow import keras
from keras.models import Sequential

model=Sequential()
model.add(keras.Input(shape=(x_train.shape[1],))) # Use Input layer to define input shape
model.add(keras.layers.Dense(128,activation='relu'))
model.add(keras.layers.Dense(64,activation='relu'))
model.add(keras.layers.Dropout(0.3)) # Moved dropout before the final Dense layer
model.add(keras.layers.Dense(1,activation='linear')) # Changed to 'linear' for regression
```

```python
model.compile(optimizer='adam',loss='mean_squared_error',metrics=['mean_absolute_error'])
model.fit(x_train,y_train,epochs=10,batch_size=64)
y_pred=model.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error
mse=mean_squared_error(y_test,y_pred)
mae=mean_absolute_error(y_test,y_pred)
print("MSE:",mse)
print("MAE:",mae)
```
